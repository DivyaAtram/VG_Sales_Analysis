```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
df= pd.read_csv("../Datasets/vgsales.csv")
df
```

```python
df.isnull().sum()
```

```python
df.Year=df.Year.fillna(0)	
```

```python
df.Publisher=df.Publisher.fillna(0)	
```

```python
df.dropna(subset=['Name','Genre','Platform'], inplace=True)
```

```python
df["Year"] = pd.to_datetime(df["Year"])
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.describe()
```

```python
df.groupby('Platform')['Global_Sales'].agg(['sum','mean'])
```

```python
df.groupby('Genre')['Global_Sales'].agg(['sum','mean'])
```

```python
df[['NA_Sales','EU_Sales','JP_Sales','Other_Sales']].agg(['sum','mean'])
```

```python
sb.histplot(df.Global_Sales)
plt.show()
```

```python
sb.boxplot(df.Global_Sales)
plt.show()
```

```python
df.sort_values(by='Global_Sales', ascending=False).head(10)
```

```python
df.Global_Sales.sum()
```

```python
df.groupby('Platform')['Global_Sales'].sum()
```

```python
Platform_Sales=df.groupby('Platform')['Global_Sales'].sum()
```

```python
sb.barplot(x=Platform_Sales.index,y=Platform_Sales.values)
plt.xticks(rotation=90)
plt.show()
```

```python
Genre_Sales=df.groupby('Genre')['Global_Sales'].sum()
```

```python
Genre_Sales
```

```python
sb.barplot(x=Genre_Sales.index,y=Genre_Sales.values)
plt.xticks(rotation=90)
plt.show()
```

```python
year_genre= df.groupby(['Year','Genre'])['Global_Sales'].sum().reset_index()
```

```python
sb.lineplot(data=year_genre, x='Year', y='Global_Sales',hue='Genre')
plt.show()
```

```python
Publisher_Sales= df.groupby('Publisher')['Global_Sales'].sum()
```

```python
top10=Publisher_Sales.sort_values(ascending=False).head(10)
sb.barplot(x=top10.index, y=top10.values)
plt.xticks(rotation=90)
plt.show()
```
